```python
purity = float(input("Enter the product purity (%): "))

if purity >= 99:
    print("Grade A")
elif purity >= 96:
    print("Grade B")
elif purity >= 90:
    print("Grade C")
else:
    print("Rejected")
```

```python
purity = float(input("Enter purity (%): "))
ph = float(input("Enter pH: "))

if purity >= 98 and 6.5 <= ph <= 7.5:
    print("Grade A")
elif purity >= 95 and 6 <= ph <= 8:
    print("Grade B")
else:
    print("Rejected")
    
```

```python
mass1 = float(input("Enter the first mass: "))
mass2 = float(input("Enter the second mass: "))

print("Addition:", mass1 + mass2)
print("Subtraction:", mass1 - mass2)
print("Multiplication:", mass1 * mass2)

if mass2 != 0:
    print("Division:", mass1 / mass2)
    print("Modulo:", mass1 % mass2)
else:
    print("Division and modulo are undefined when the second mass is zero.")
    
```

```python
import math

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

print("Quadratic equation: {}x^2 + {}x + {}".format(a, b, c))

if a == 0:
    print("It is not a quadratic equation")

else:
    D = b**2 - 4*a*c

    if D > 0:
        root1 = (-b + math.sqrt(D)) / (2*a)
        root2 = (-b - math.sqrt(D)) / (2*a)

        print("Two distinct real roots")
        print("Root 1 =", root1)
        print("Root 2 =", root2)

    elif D == 0:
        root = -b / (2*a)
        print("Two equal real roots")
        print("Root =", root)

    else:
        real = -b / (2*a)
        imaginary = math.sqrt(-D) / abs(2*a)

        print("Complex roots")
        print("Root 1 =", real, "+", imaginary, "i")
        print("Root 2 =", real, "-", imaginary, "i")
```
