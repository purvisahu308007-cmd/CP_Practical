```python
# create a simple if elif else code to check marks and give 
# grade A , B, C, D, f based on the marks entered by the user.
marks = float(input('Enter your marks: '))

if marks >= 95:
    grade = 'A'
elif marks >= 90:
    grade = 'B'
elif marks >= 80:
    grade = 'C'
elif marks >= 70:
    grade = 'D'
else:
    grade = 'F'

print(f'Hello {name}, your grade is {grade}.')
```
